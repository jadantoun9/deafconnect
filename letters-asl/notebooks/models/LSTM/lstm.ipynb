```python
import tensorflow as tf
import numpy as np
```

### 1) Loading data

```python
landmarks_data = np.load('../../../data/divided/landmarks_dataset.npz')

# Extract the individual arrays using their keys
X_train_landmarks = landmarks_data['X_train']
y_train_landmarks = landmarks_data['y_train']
X_test_landmarks = landmarks_data['X_test']
y_test_landmarks = landmarks_data['y_test']

images_data = np.load('../../../data/divided/images_dataset.npz')

# Extract the individual arrays using their keys
X_train_images = images_data['X_train']
y_train_images = images_data['y_train']
X_test_images = images_data['X_test']
y_test_images = images_data['y_test']
```

### 2) Defining LSTM Model Architecture

```python
# Define LSTM Model
def build_lstm(input_shape, num_classes):
    model = tf.keras.Sequential([
        tf.keras.layers.LSTM(128, input_shape=input_shape),
        tf.keras.layers.Dense(64, activation='relu'),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    return model
```

### 3) Creating an LSTM model trained on hand landmarks

```python
# Reshape landmarks to fit LSTM's input shape
input_shape = X_train_landmarks.shape[1:]  # Shape of one sample
num_classes = len(np.unique(y_train_landmarks))  # Number of unique labels

# Build the LSTM model
lstm_landmarks_model = build_lstm(input_shape, num_classes)

# Compile the LSTM model
lstm_landmarks_model.compile(optimizer='adam',
                   loss='sparse_categorical_crossentropy',
                   metrics=['accuracy'])

# Train the LSTM model
lstm_landmarks_model.fit(X_train_landmarks, y_train_landmarks, epochs=10, batch_size=32)
```

```python
# Evaluate the model on the test data
test_loss, test_accuracy = lstm_landmarks_model.evaluate(X_test_landmarks, y_test_landmarks)
print(f'accuracy {test_accuracy}')
```

#### Saving model for persistance

```python
lstm_landmarks_model.save('../../../trained-models/lstm_landmarks_model.h5')
```

```python
def resize_images(X):
    resized_X = []
    for image in X:
        resized_image = cv2.resize(image, (64, 64))
        resized_X.append(resized_image)
    return np.array(resized_X)
```

### 5) Data Augmentation

```python
def apply_augmentation(image):
    # Randomly choose augmentation techniques
    if np.random.uniform() < 0.5:
        # Randomly rotate the image by a degree between -20 and 20
        angle = np.random.uniform(low=-20, high=20)
        rows, cols = image.shape[:2]
        rotation_matrix = cv2.getRotationMatrix2D((cols / 2, rows / 2), angle, 1)
        image = cv2.warpAffine(image, rotation_matrix, (cols, rows))

    if np.random.uniform() < 0.5:
        # Randomly shift the image horizontally and vertically by 10% of its width/height
        rows, cols = image.shape[:2]
        x_shift = np.random.uniform(low=-0.1 * cols, high=0.1 * cols)
        y_shift = np.random.uniform(low=-0.1 * rows, high=0.1 * rows)
        translation_matrix = np.float32([[1, 0, x_shift], [0, 1, y_shift]])
        image = cv2.warpAffine(image, translation_matrix, (cols, rows))

    if np.random.uniform() < 0.5:
        # Randomly flip the image horizontally
        image = cv2.flip(image, 1)

    if np.random.uniform() < 0.5:
        # Randomly flip the image vertically
        image = cv2.flip(image, 0)

    return image
```

```python
# Visualize images before and after data augmentation
num_images_to_visualize = 5

for i in range(num_images_to_visualize):
    # Randomly select an image index
    idx = np.random.randint(0, len(X))

    # Original image
    original_image = X_scaled[idx]
    
    # Augmented image
    augmented_image = apply_augmentation(original_image)

    # Plotting
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(original_image)
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(augmented_image)
    plt.title('Augmented Image')
    plt.axis('off')

    plt.show()
```

```python
# Apply augmentation to selected images
for i in range(len(X_scaled)):
    # Randomly choose whether to apply augmentation
    if np.random.uniform() < 0.5:
        X_scaled[i] = apply_augmentation(X_scaled[i])
```

### 6) Extract Hand Landmarks using MediaPipe

```python
def get_hand_landmarks(results, y, X_landmarks, y_landmarks):

    if results.multi_hand_landmarks:
        # Since only 1 hand is allowed
        hand = results.multi_hand_landmarks[0]
        hand_landmarks = []
        for landmark in hand.landmark:
            hand_landmarks.append([landmark.x, landmark.y, landmark.z]) 
        
        X_landmarks.append(hand_landmarks)
        y_landmarks.append(y)
```

```python
# Initialize MediaPipe Hands
mp_hands = mp.solutions.hands 

# Mediapipe Hands model
hands = mp_hands.Hands(static_image_mode=False, max_num_hands=1, min_detection_confidence=0.5)
```

```python
X_landmarks = []
y_landmarks = []

count = 0
for image, label in zip(X, y):
    # Process image to get hand landmarks
    count+=1
    print(count)
    results = hands.process(image)
    get_hand_landmarks(results, label, X_landmarks, y_landmarks)
```

```python
# Define the function to draw hand landmarks on the image
def draw_hand_landmarks(image, results):
    mp_drawing = mp.solutions.drawing_utils
    # Draw landmarks on the image
    if results.multi_hand_landmarks:
        annotated_image = image.copy()  # Make a copy of the image to draw landmarks on
        for hand_landmarks in results.multi_hand_landmarks:
             mp_drawing.draw_landmarks(image=annotated_image, landmark_list=hand_landmarks,
                                        connections=mp.solutions.hands.HAND_CONNECTIONS)
        return annotated_image
    else:
        return image

# Collect annotated images
annotated_images = []

for i in range(15):
    results = hands.process(X[i])
    annotated_image = draw_hand_landmarks(X[i], results)
    annotated_images.append(annotated_image)

# Display images next to each other
num_images = len(annotated_images)
num_cols = 5  # Number of columns per row

num_rows = (num_images + num_cols - 1) // num_cols  # Calculate the number of rows needed

plt.figure(figsize=(num_cols * 2, num_rows * 2))

for i in range(num_images):
    plt.subplot(num_rows, num_cols, i + 1)
    plt.imshow(cv2.cvtColor(annotated_images[i], cv2.COLOR_BGR2RGB))
    plt.title('Image {}'.format(i + 1))
    plt.axis('off')

plt.tight_layout()  # Adjust layout to prevent overlap
plt.show()
```

```python
X_landmarks = np.array(X_landmarks)
y_landmarks = np.array(y_landmarks)
```

```python
# Create a table using pandas to count the occurrences of each letter
import pandas as pd
def show_y_table(y):
    letter_counts = pd.Series(y).value_counts().reset_index()
    letter_counts.columns = ['Letter', 'Count']
    letter_counts = letter_counts.sort_values(by='Letter')
    return letter_counts
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
```

### 1) Dividing Landmarks data into training and testing

```python
# Load the compressed .npz file
landmarks_data = np.load('../../data/processed/landmarks.npz')

# Extract the individual arrays using their keys
X_landmarks = landmarks_data['X']
y_landmarks = landmarks_data['y']
```

```python
X_landmarks.shape
```

```python
y_landmarks.shape
```

```python
# Stratified 70/15/15 split: train, val, test.
# random_state=42 keeps the original train/test boundaries reproducible.
X_tmp, X_test_landmarks, y_tmp, y_test_landmarks = train_test_split(
    X_landmarks, y_landmarks, test_size=0.15, random_state=42, stratify=y_landmarks,
)
X_train_landmarks, X_val_landmarks, y_train_landmarks, y_val_landmarks = train_test_split(
    X_tmp, y_tmp, test_size=0.15/0.85, random_state=42, stratify=y_tmp,
)
print('landmarks shapes — train', X_train_landmarks.shape, 'val', X_val_landmarks.shape, 'test', X_test_landmarks.shape)
```

#### Saving data for persistance

```python
np.savez_compressed(
    "../../data/divided/landmarks_dataset.npz",
    X_train=X_train_landmarks,
    y_train=y_train_landmarks,
    X_val=X_val_landmarks,
    y_val=y_val_landmarks,
    X_test=X_test_landmarks,
    y_test=y_test_landmarks,
)
```

### 2) Dividing Images data into training and testing

```python
# Load the compressed .npz file
landmarks_data = np.load('../../data/processed/images.npz')

# Extract the individual arrays using their keys
X_images = landmarks_data['X']
y_images = landmarks_data['y']
```

```python
X_images.shape
```

```python
y_images.shape
```

```python
X_tmp_i, X_test_images, y_tmp_i, y_test_images = train_test_split(
    X_images, y_images, test_size=0.15, random_state=42, stratify=y_images,
)
X_train_images, X_val_images, y_train_images, y_val_images = train_test_split(
    X_tmp_i, y_tmp_i, test_size=0.15/0.85, random_state=42, stratify=y_tmp_i,
)
print('images shapes — train', X_train_images.shape, 'val', X_val_images.shape, 'test', X_test_images.shape)
```

```python
# Now, you have your training and testing sets
print("X_train_images shape:", X_train_images.shape) 
print("y_train_images shape:", y_train_images.shape)  

print("X_test_images shape:", X_test_images.shape)   
print("y_test_images shape:", y_test_images.shape) 
```

```python
np.savez_compressed(
    "../../data/divided/images_dataset.npz",
    X_train=X_train_images,
    y_train=y_train_images,
    X_val=X_val_images,
    y_val=y_val_images,
    X_test=X_test_images,
    y_test=y_test_images,
)
```
